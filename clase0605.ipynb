```python
def procesar(files: list, path: str) -> pd.DataFrame:
    corpus = {'doc_id': [], 'raw': [], 'processed': [], 'tf_idf': []}
    stemmer_espanol = SnowballStemmer('spanish')
    for doc_id, file in enumerate(files):
        with open(path + file, 'r', encoding='utf-8') as f:
            text = f.read()
            clean_text = re.sub(r'[^\w\s]', '', text)
            stemmed_words = [stemmer_espanol.stem(word.lower()) for word in clean_text.split()]    
            corpus['doc_id'].append(doc_id)
            corpus['raw'].append(text)
            corpus['processed'].append(' '.join(stemmed_words))
    return pd.DataFrame(corpus)
```

```python
df_corpus = procesar(files, path)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(df_corpus['processed'])
feature_names = tfidf_vectorizer.get_feature_names_out()
```
